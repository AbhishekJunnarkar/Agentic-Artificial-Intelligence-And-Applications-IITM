### if and else

```python
x = 10
if x> 5:
    print('X is greater than 5')
```

```python
# Correct logic example of if, elif and else
score = 90
if score >= 90:
    print('Excellent')
elif score >=80:
    print('V. Good')
else:
    print('Good')
```

```python
# Incorrect logic exmaple
score = 90
if score >= 90:
    print('Excellent')
if score >=80:
    print('V. Good')
if score >=70: 
    print('Good')
```

```python
coin_toss = 'Head'
if coin_toss == 'Head':
    print('Its a head')
elif coin_toss == 'Tail':
    print('Its a Tail')
```

### For

```python
for i in range(1,11):
    print(i)
```

```python
stud_list = ['Arnav', 'Aarya', 'Himani']
for stud in stud_list:
    print(stud)
```

```python
stud_map = {'name':'Arnav', 'age' : 13, 'gender' : 'M'}
for stud in stud_map:
    print(f'{stud}')
    print(type(stud))
for stud in stud_map.values():
    print(f'{stud}')
    print(type(stud))
for stud in stud_map.items():
    print(f'{stud}')
    print(type(stud))
```
